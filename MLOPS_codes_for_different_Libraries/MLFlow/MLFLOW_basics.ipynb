# What is Experiment Tracking

Experiment tracking is a key component of the machine learning lifecycle that involves recording and organizing the details of machine learning experiments. This includes tracking parameters, metrics, artifacts, and other relevant information to facilitate reproducibility, collaboration, and analysis of model performance.

# MLflow

## What it is (in simple terms)

Imagine you're baking many versions of a cake, changing the flour, sugar, and oven temperature each time. Without notes, you'd forget which recipe gave the best result. **MLflow is a notebook for machine learning experiments.** It records every training attempt: the settings you used, the scores you got, and the model file you produced. It also helps you find the best attempt later and deploy it.

It's open-source and works with most ML libraries, such as scikit-learn, TensorFlow/Keras, and PyTorch.

## Main Components

| Component | What it does | Simple analogy |
|---|---|---|
| **Tracking** | Logs parameters, metrics, code versions, and output files for each experiment run | A lab notebook |
| **Projects** | Packages your ML code with its dependencies so others can run it the same way | A recipe card with a shopping list |
| **Models** | Saves a trained model in a standard format that can be loaded or served anywhere | A sealed, labeled container |
| **Model Registry** | Stores models centrally with version numbers and lifecycle stages (e.g., staging, production) | A catalogue or library of approved models |

### Brief details

- **Tracking**
  - *Experiments* group related runs.
  - *Runs* are single training attempts.
  - *Parameters* are inputs such as learning rate. *Metrics* are results such as accuracy. *Artifacts* are files such as plots or model weights.
  - Autologging can capture these automatically for supported libraries.
- **Projects**: Defined by an `MLproject` file that specifies the environment, entry points, and parameters.
- **Models**: Uses a "flavor" system, so one saved model can be loaded by different tools (e.g., as a Python function or a REST endpoint).
- **Model Registry**: Tracks which model version is current and who promoted it, which supports governance.

MLflow has also added components for newer workflows, such as evaluation and tracing for LLM applications. Check the current docs for details, since the project changes often.

## Where it fits

A typical flow looks like this:

1. Train several models with different settings and log each run with **Tracking**.
2. Compare runs in the MLflow UI and pick the best one.
3. Save it as a **Model** and register it in the **Model Registry**.
4. Deploy the registered version.
